```python
#building and reading a Series
s = pd.Series([72, 85, 90, 66], index=["Asha", "Ravi", "Meera", "Kabir"], name="marks")
print(s, "\n")
print(type(s).__name__, "| dtype", s.dtype, "| shape", s.shape, "| name", s.name)
print("index :", s.index.tolist())
print("values:", s.values, type(s.values).__name__)             # a NumPy array underneath

print(s["Ravi"], s.loc["Ravi"], s.iloc[1])                       # label, label, position
print(s["Ravi":"Meera"].tolist(), "<- a LABEL slice includes the end")
print(s.iloc[1:3].tolist(), "<- a POSITION slice excludes the end")
print(s[["Asha", "Kabir"]].tolist())
```

```python
#ways to create a Series
print(pd.Series([10, 20, 30]), "\n")                               # the default index is 0, 1, 2 (a RangeIndex)
print(pd.Series({"a": 1, "b": 2, "c": 3}), "\n")                   # from a dictionary: keys become the index
print(pd.Series(5, index=["x", "y", "z"]), "\n")                   # a scalar repeated
print(pd.Series(np.arange(3) ** 2, index=list("pqr")), "\n")       # from a NumPy array

print(pd.Series([1, 2, None]).tolist(), pd.Series([1, 2, None]).dtype)     # a missing value turns ints into floats
print(pd.Series(["a", "b"]).dtype, "|", pd.Series([1, "a", 2.5]).dtype)    # text and mixed data
print(pd.Series([1, 2, None], dtype="Int64").tolist())                     # the nullable integer type keeps integers
```

```python
#vector operations and the descriptive statistics you know
print((s * 2).tolist(), (s > 70).tolist())
print(s[s > 70])                                                 # a boolean mask (as in NumPy)
print("mean", s.mean(), "| median", s.median(), "| sample SD", round(s.std(), 3), "(pandas uses ddof=1)")
print("max", s.max(), "at label", s.idxmax(), "| min", s.min(), "at label", s.idxmin())   # idxmax gives the LABEL, argmax the position
print(s.sort_values(ascending=False).head(2).to_dict())
print(s.rank(ascending=False).to_dict())
print(s.describe())

colors = pd.Series(["red", "blue", "red", "green", "red"])
print(colors.value_counts().to_dict(), colors.unique(), colors.nunique())
print(s.between(70, 90).tolist(), s.isin([72, 66]).tolist())
```

```python
#alignment by label
a = pd.Series([1, 2, 3], index=["a", "b", "c"])
b = pd.Series([10, 20, 30], index=["b", "c", "d"])
print(a + b, "\n")                                                  # matched by LABEL: a and d have no partner, so they become NaN
print(a.add(b, fill_value=0), "\n")                                 # treat a missing partner as 0
print(np.array([1, 2, 3]) + np.array([10, 20, 30]))                 # NumPy combines by POSITION
```

```python
#at and iat: one value, fast
# scalar access, and what it costs
print(by_name.at["Asha", "marks"], students.iat[0, 3], "<- one value by label (at) and by position (iat)")

d = by_name.copy()
d.at["Asha", "marks"] = 74.0
d.iat[1, 2] = 86.0                                                       # row 1 (Ravi), column 2 (marks)
print(d["marks"].head(2).to_dict(), "| original untouched:", by_name["marks"].head(2).to_dict())

frame = pd.DataFrame(rng.random((2000, 4)), columns=list("abcd"))

def best_time(f, repeat=3):
    best = float("inf")
    for _ in range(repeat):
        start = time.perf_counter()
        f()
        best = min(best, time.perf_counter() - start)
    return best

for label, f in (("loc ", lambda: [frame.loc[i, "a"] for i in range(2000)]),
                 ("at  ", lambda: [frame.at[i, "a"] for i in range(2000)]),
                 ("iloc", lambda: [frame.iloc[i, 0] for i in range(2000)]),
                 ("iat ", lambda: [frame.iat[i, 0] for i in range(2000)]),
                 ("whole column at once", lambda: frame["a"].tolist())):
    print(f"{label:<22}{best_time(f) * 1000:8.2f} ms for 2000 values")
```

```python
#Mini project, load and explore a sales file
#create a file
n = 200
cities = ["Delhi", "Mumbai", "Pune", "Chennai"]
prices = {"Notebook": 45.5, "Pen": 8.25, "Bag": 1299.0, "Marker": 32.0}
sales_raw = pd.DataFrame({
    "order_id": np.arange(1001, 1001 + n),
    "order_date": pd.date_range("2026-09-01", periods=n, freq="8h").strftime("%Y-%m-%d"),
    "city": rng.choice(cities, n, p=[0.4, 0.3, 0.2, 0.1]),
    "product": rng.choice(list(prices), n),
    "qty": rng.integers(1, 12, n).astype(float),
})
sales_raw["unit_price"] = sales_raw["product"].map(prices)              # map is covered on S11
sales_raw.loc[rng.choice(n, 8, replace=False), "qty"] = np.nan             # a few missing quantities
sales_raw.to_csv(SCRATCH / "sales.csv", index=False)
print((SCRATCH / "sales.csv").read_text(encoding="utf-8").splitlines()[:4])
```
